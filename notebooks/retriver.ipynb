```python
from src.retriever import QdrantIndexer
from src.utils.config import get_config 
```

```python
cfg = get_config()
indexor = QdrantIndexer()
```

```python
responses = indexor.retriever('Give me apple microphone under 100 dollars', 5)
```

```python
results = []

for point in responses.points:

    payload = point.payload or {}

    results.append(
        {
            "id": point.id,
            "score": point.score,
            "parent_asin": payload.get(
                "parent_asin"
            ),
            "document": payload.get(
                "document"
            ),
            "metadata": payload.get(
                "metadata"
            ),
        }
    )
```
