```python
from src.reranking import CrossEncoderRanker
from src.retriever import QdrantIndexer
```

```python
indexor = QdrantIndexer()
```

```python
reranker = CrossEncoderRanker()
```

```python
query = 'Give me apple microphone under 100 dollars'
responses = indexor.retriever(query, 40)
```

```python
results = []

for point in responses.points:

    payload = point.payload or {}

    results.append(
        {
            "id": point.id,
            "score": point.score,
            "parent_asin": payload.get(
                "parent_asin"
            ),
            "document": payload.get(
                "document"
            ),
            "metadata": payload.get(
                "metadata"
            ),
        }
    )
```

```python
reranked_results = reranker.rerank(query, results)
```
