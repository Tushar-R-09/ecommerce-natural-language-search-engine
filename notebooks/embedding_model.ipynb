```python
from sentence_transformers import SentenceTransformer
```

```python
model = SentenceTransformer("BAAI/bge-base-en-v1.5")
```

```python
sentences = [
    "A dog is playing in the park",
    "A puppy is running outside.",
    "A cat is sleeping on the sofa",
    "The stock market crashed today"
]
```

```python
embeddings = model.encode(
    sentences,
    normalize_embeddings = True
)
```

```python
embeddings.shape
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity_matrix = cosine_similarity(embeddings)
```

```python
similarity_matrix
```

```python
documents = [
    "Wireless noise cancelling headphones with deep bass",
    "Bluetooth earbuds with active noise cancellation",
    "Men's running shoes with breathable mesh",
    "Gaming keyboard with RGB backlighting"
]

query = "headphones for listening to music"
```

```python
document_embeddings = model.encode(
    documents,
    normalize_embeddings=True
)
```

```python
query_embedding = model.encode(
    [query],
    normalize_embeddings=True
)
```

```python
document_embeddings.shape
```

```python
scores = cosine_similarity(
    query_embedding,
    document_embeddings
)[0]
```

```python
results = sorted(
    zip(documents, scores),
    key=lambda x: x[1],
    reverse=True
)
```

```python
print(f"Query: {query}\n")
for rank, (document, score) in enumerate(results, 1):
    print(f"{rank}. {score:.4f} {document}")
```
